# Winds at FL430 (162.4 hPa), Dec/Jan/Feb/Mar 2015–2025

How often is the flow at release height equatorward? For each month: maps of the mean meridional wind,
its day-to-day variability, the fraction of days with v < 0 (poleward in the SH), the e-folding
autocorrelation timescale of daily-mean v, and the mean zonal wind.

Source: ERA5 via the public ARCO-ERA5 zarr on Google Cloud. u and v at 150 and 175 hPa are interpolated
in ln p to 162.4 hPa and stored 6-hourly, one file per calendar month and year (stage 1, resumable).
Daily means are formed in stage 2. Needs `gcsfs` and `zarr`.

**e-folding time τ.** For each grid point and month, r₁ is the correlation between daily-mean v on
consecutive days (anomalies from that month's mean at that point; pairs taken within each month and pooled
over the 11 years). If v behaves like a first-order autoregressive process, the autocorrelation at lag k
days is r₁ᵏ = exp(−k/τ) with τ = −1/ln r₁. τ is the number of days over which the correlation decays to
1/e ≈ 0.37, i.e. how long today's meridional wind remains informative about the wind ahead. τ ≈ 1 day
(r₁ ≈ 0.37) means successive days are nearly independent and equatorward spells are short; τ of several
days means spells persist. The AR(1) form is an approximation; τ is set to NaN where r₁ ≤ 0.

```python
import glob
import os
import time

import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.path as mpath
import cartopy.crs as ccrs
from cartopy.util import add_cyclic_point

# ---- configuration ----
ARCO_ERA5 = "gs://gcp-public-data-arco-era5/ar/full_37-1h-0p25deg-chunk-1.zarr-v3"
YEARS = (2015, 2025)               # inclusive
MONTHS = [12, 1, 2, 3]             # row order in the figure
MONTH_NAMES = {12: "Dec", 1: "Jan", 2: "Feb", 3: "Mar"}
LAT_MAX = -30.0
HOURS = [0, 6, 12, 18]
P_FL430 = 226.32 * np.exp(-9.80665 * (43000 * 0.3048 - 11000.0) / (287.05287 * 216.65))   # 162.4 hPa
WIND_DIR = "winds_fl430"           # stage-1 output, one file per year-month
STATS_FILE = f"winds_fl430_stats_DJFM_{YEARS[0]}-{YEARS[1]}.nc"
FIG_FILE = f"winds_fl430_DJFM_{YEARS[0]}-{YEARS[1]}.pdf"
os.makedirs(WIND_DIR, exist_ok=True)
print(f"FL430 = {P_FL430:.2f} hPa")
```

## Stage 1: extract u and v at FL430 from ARCO-ERA5

Reads `u_component_of_wind` and `v_component_of_wind` at 150 and 175 hPa for every day of each month at
`HOURS`, selects the 1° grid points (lat ≤ −30°, lon −180…179), interpolates in ln p to 162.4 hPa, and
writes one file per year-month. Not yet verified against the live store: variable names, coordinate names
and how far the store extends in time. The cell prints the store's time range; a month beyond the end
(December 2025 is the most likely) is skipped and reported.

```python
def wind_path(y, m):
    return os.path.join(WIND_DIR, f"uv_fl430_{y}-{m:02d}.nc")


need = [(y, m) for y in range(YEARS[0], YEARS[1] + 1) for m in MONTHS if not os.path.exists(wind_path(y, m))]
print("year-months to fetch:", need)

if need:
    era = xr.open_zarr(ARCO_ERA5, chunks=None, storage_options=dict(token="anon"))
    print("time range:", str(era.time.values[0])[:10], "..", str(era.time.values[-1])[:10])
    lat_j = np.arange(-90.0, LAT_MAX + 0.5, 1.0)
    lon_j = np.arange(-180.0, 180.0, 1.0)
    lon_e = lon_j % 360.0                              # ARCO longitudes run 0..359.75
    w175 = (np.log(P_FL430) - np.log(150.0)) / (np.log(175.0) - np.log(150.0))
    for y, m in need:
        days = pd.date_range(f"{y}-{m:02d}-01", periods=pd.Period(f"{y}-{m:02d}").days_in_month, freq="D")
        times = np.array([np.datetime64(d.replace(hour=h)) for d in days for h in HOURS])
        if times[-1] > era.time.values[-1]:
            print(f"  {y}-{m:02d}: beyond end of store, skipped")
            continue
        t0 = time.time()
        sub = (era[["u_component_of_wind", "v_component_of_wind"]]
               .sel(time=times, level=[150, 175])
               .sel(latitude=lat_j, longitude=lon_e, method="nearest").load())
        sub = sub.assign_coords(longitude=lon_j).rename(latitude="lat", longitude="lon")
        out = xr.Dataset()
        for src, name in [("u_component_of_wind", "u"), ("v_component_of_wind", "v")]:
            f = (1 - w175) * sub[src].sel(level=150) + w175 * sub[src].sel(level=175)
            out[name] = f.drop_vars("level", errors="ignore").astype("f4")
            out[name].attrs = dict(units="m s-1", long_name=f"{name} at {P_FL430:.1f} hPa (FL430), ln-p interpolated 150/175 hPa")
        out.attrs = dict(source=ARCO_ERA5, hours=str(HOURS))
        out.to_netcdf(wind_path(y, m))
        print(f"  {y}-{m:02d}: {out.sizes['time']} steps, v range {float(out.v.min()):.1f}..{float(out.v.max()):.1f} m/s, "
              f"{time.time() - t0:.0f} s")

uv = xr.concat([xr.load_dataset(f) for f in sorted(glob.glob(os.path.join(WIND_DIR, "*.nc")))], "time").sortby("time")
print(uv.sizes)
print("steps per (year, month):")
print(uv.time.to_series().groupby([uv.time.dt.year.values, uv.time.dt.month.values]).size().unstack(1)[MONTHS].fillna(0).astype(int))
```

## Stage 2: daily means and statistics, per month

- daily-mean u and v from the 6-hourly values
- mean and standard deviation of daily v over all days of the month, all years
- fraction of days with v < 0 (poleward)
- lag-1 autocorrelation r₁ of daily-v anomalies (grid-point monthly mean removed), from consecutive-day
  pairs within each month, pooled over years; τ = −1/ln r₁ (NaN where r₁ ≤ 0)
- mean u

```python
daily = uv.resample(time="1D").mean()
daily = daily.sel(time=daily.time.dt.month.isin(MONTHS))
years = sorted(set(daily.time.dt.year.values))

per_month = []
for m in MONTHS:
    dm = daily.sel(time=daily.time.dt.month == m)
    v, u = dm.v, dm.u
    v_mean = v.mean("time")
    a = v - v_mean
    num = xr.zeros_like(v_mean)
    den = xr.zeros_like(v_mean)
    for y in sorted(set(v.time.dt.year.values)):
        ay = a.sel(time=str(y)).values                   # (nday, nlat, nlon), one calendar month
        num += (ay[:-1] * ay[1:]).sum(0)
        den += (ay ** 2).sum(0)
    r1 = num / den
    tau = -1.0 / np.log(r1.where(r1 > 0))
    st = xr.Dataset(dict(v_mean=v_mean, v_std=v.std("time", ddof=1), frac_poleward=(v < 0).mean("time"),
                         r1=r1, tau=tau, u_mean=u.mean("time"), n_days=v.sizes["time"]))
    per_month.append(st.expand_dims(month=[m]))
    print(f"{MONTH_NAMES[m]}: {v.sizes['time']} days over {len(set(v.time.dt.year.values))} years")

stats = xr.concat(per_month, "month")
stats.attrs = dict(years=f"{years[0]}-{years[-1]}", p_hPa=float(P_FL430), hours=str(HOURS))
stats.to_netcdf(STATS_FILE)

print("\nzonal means by latitude band (v_mean m/s | P(v<0) | tau d | u_mean m/s):")
bands = [(-40, -30), (-50, -40), (-60, -50), (-70, -60), (-80, -70), (-90, -80)]
for lo, hi in bands:
    s_ = stats.sel(lat=slice(lo, hi - 1e-6))
    print(f"  {-hi:2d}-{-lo:2d}S: " + " | ".join(
        f"{MONTH_NAMES[m]} {float(s_.v_mean.sel(month=m).mean()):+.1f} {float(s_.frac_poleward.sel(month=m).mean()):.2f} "
        f"{float(s_.tau.sel(month=m).mean(skipna=True)):.1f} {float(s_.u_mean.sel(month=m).mean()):+.0f}" for m in MONTHS))
```

## Maps

One row per month; columns: mean v, standard deviation of daily v, fraction of days with v < 0, τ, mean u.
South polar view, 30–90°S, latitude circles at 45, 60 and 75°S. Colour scales are shared down each column.
Black contours: v = 0 and u = 0 on the mean-wind maps, 0.5 on the fraction map.

```python
stats = xr.open_dataset(STATS_FILE)
lat, lon = stats.lat.values, stats.lon.values
theta = np.linspace(0, 2 * np.pi, 200)
circle = mpath.Path(np.vstack([np.sin(theta), np.cos(theta)]).T * 0.5 + 0.5)
pc = ccrs.PlateCarree()
lon_line = np.linspace(-180, 180, 361)

vlim = float(np.ceil(abs(stats.v_mean).quantile(0.99)))
ulim = float(np.ceil(abs(stats.u_mean).quantile(0.99)))
cols = [("v_mean", "mean v [m s⁻¹]  (>0 equatorward)", "RdBu_r", -vlim, vlim, 0.0),
        ("v_std", "std of daily v [m s⁻¹]", "viridis", 0, float(np.ceil(stats.v_std.quantile(0.99))), None),
        ("frac_poleward", "fraction of days with v < 0", "cividis", 0, 1, 0.5),
        ("tau", "e-folding time τ [days]", "magma", 0, float(np.ceil(stats.tau.quantile(0.98))), None),
        ("u_mean", "mean u [m s⁻¹]", "RdBu_r", -ulim, ulim, 0.0)]

nrow, ncol = len(MONTHS), len(cols)
fig, axs = plt.subplots(nrow, ncol, figsize=(3.3 * ncol, 2.95 * nrow + 0.6), squeeze=False,
                        subplot_kw=dict(projection=ccrs.SouthPolarStereo()))
ims = {}
for i, m in enumerate(MONTHS):
    for j, (var, label, cmap, vmin, vmax, clev) in enumerate(cols):
        ax = axs[i, j]
        ax.set_extent([-180, 180, -90, LAT_MAX], pc)
        ax.set_boundary(circle, transform=ax.transAxes)
        data, lonc = add_cyclic_point(stats[var].sel(month=m).values, coord=lon)
        ims[j] = ax.pcolormesh(lonc, lat, data, cmap=cmap, vmin=vmin, vmax=vmax, shading="auto",
                               transform=pc, rasterized=True)
        ax.coastlines(lw=0.4, color="0.3")
        if clev is not None:
            ax.contour(lonc, lat, data, levels=[clev], colors="k", linewidths=0.6, transform=pc)
        for la in (-45, -60, -75):
            ax.plot(lon_line, np.full_like(lon_line, la), transform=pc, color="0.85", lw=0.5, ls="--", zorder=3)
            ax.text(135, la, f"{-la}°S", transform=pc, fontsize=6, color="0.2", ha="center", va="center", zorder=4,
                    bbox=dict(boxstyle="round,pad=0.1", fc="w", ec="none", alpha=0.7))
        if i == 0:
            ax.set_title(label, fontsize=9.5)
        if j == 0:
            ax.text(-0.08, 0.5, MONTH_NAMES[m], transform=ax.transAxes, rotation=90, va="center", ha="center", fontsize=12)
for j, (var, label, *_ ) in enumerate(cols):
    fig.colorbar(ims[j], ax=axs[:, j], orientation="horizontal", fraction=0.025, pad=0.02, aspect=22)
fig.suptitle(f"Winds at FL430 ({P_FL430:.0f} hPa), daily means, {stats.attrs['years']} (ERA5)", y=0.995)
fig.savefig(FIG_FILE, bbox_inches="tight", dpi=200)
print("wrote", FIG_FILE)
plt.show()
```
